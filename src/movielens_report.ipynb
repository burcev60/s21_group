Создадим экземлпяры классов для дальнейшего анализа фильмов

```python
links = Links("../data/links.csv")
movies = Movies("../data/movies.csv")
tags = Tags("../data/tags.csv")
```

Анализ тегов, посмотрим что у нас тут происходит.

Для начала оценим какие у нас есть самые большие теги, которые оставили пользователи и посмотрим сколько в них содержится слов

```python
%%timeit -r1 -n1
most_words = tags.most_words(5)
for tag, words_count in most_words.items():
    print(f"{tag} : {words_count}")
```

А вдруг пользователи оставляли в тегах много коротких слов? Посмотрим теперь на длину по количеству символов.

```python
%%timeit -r1 -n1
longest = tags.longest(5)
for tag, char_count in longest.items():
    print(f"{tag} : {char_count}")
```

Посмотрим теперь на пересечение самых длинных тегов по словам и по символам.

```python
%%timeit -r1 -n1
most_words_and_longest = tags.most_words_and_longest(5)
for tag in most_words_and_longest:
    print(tag)
```

Давайте теперь посмотрим на самые популярные теги и как часто их указывали.

```python
%%timeit -r1 -n1
most_popular = tags.most_popular(5)
for tag, count in most_popular.items():
    print(f"{tag} : {count}")
```

Интересно, есть ли такие теги, где пользователи были невероятно довольны фильмом. Попробуем найти такие теги, где пользователи поставили оценку "best" фильм!

```python
%%timeit -r1 -n1
tags_with_best = tags.tags_with("best")
for tag in tags_with_best:
    print(tag)
```

Теперь давайте посмотрим что там у нас по фильмам. Начнем с того, что посмотрим какие года были самыми плодовитыми на новинки.

```python
%%timeit -r1 -n1
dist_by_release = movies.dist_by_release()
for year, count in dist_by_release.items():
    print(f"{year} : {count}")
```

А давайте теперь посмотрим что у нас там творится на IMDB.
Мы можем получить сколько угодно информации с сайта IMDB, но давайте для демонстрации ограничимся тремя фильмами. Еще также укажем какие нам интересно узнать поля, это название, режиссер, бюжет, сборы и продолжительность.

```python
%%timeit -r1 -n1
list_of_movies = ["0317198", "0308644", "0368891"]
list_of_fields = ['Title', 'Director', 'Budget', 'Cumulative Worldwide Gross', 'Runtime']

get_imdb = links.get_imdb(list_of_movies, list_of_fields)
for film in get_imdb:
    print(*film, sep = "|")
```

Можем просто посмотреть названия и продолжительность, если поменяем список полей.

```python
%%timeit -r1 -n1
list_of_movies = ["0317198", "0308644", "0368891"]
list_of_fields = ['Title', 'Runtime']

get_imdb = links.get_imdb(list_of_movies, list_of_fields)
for film in get_imdb:
    print(*(film[1:]), sep = "|")
```

Мы заранее спарсили информацию по 1000 фильмам, чтобы считать статистику "на лету". Посмотрим на режиссеров, у которых больше всего фильмов в данных.

```python
%%timeit -r1 -n1
top_directors = links.top_directors(5)
for director, films in top_directors.items():
    print(f"{director} : {films}")
```
